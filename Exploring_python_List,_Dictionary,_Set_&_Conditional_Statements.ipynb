**List (Creation, Modification, and Access):**

**1. List Creation:**

```python
# a. Create a list named age_list with five integer elements. For eg., [24, 25, 27, 28 ,29]

age_list = [24, 25, 27, 28 ,29]

# b. Create a list named name_list with five string elements.

name_list = ['Kala', 'Mala', 'Sri', 'Vino', 'Raj']

print("Age list:", age_list)
print("Name list:", name_list)
```

**2. List Operations / Modifications:**

```python
# a. Append the string "Yazhini" to name_list.
name_list.append("Yazhini")
print("Appended Name list:", name_list)

# b. Insert the element 26 at index 2 in age_list.
age_list.insert(2,26)
print("Inserted Age list:", age_list)

# c. Remove the string "Yazhini" from name_list.
name_list.remove("Yazhini")
print("Removed Name list:", name_list)

# d. Pop the last element from age_list.
age_list.pop()
print("Poped Age list:", age_list)


# e. Extend the age_list with additional ages [31,30,32].
age_list.extend([31,30,32])
print("Extended Age list:", age_list)

# f. Sort age_list in descending order.
age_list.sort(reverse=True)
print("Descending order Age list:", age_list)

# g. Find the max age, min age, and sum of all ages from age_list.
Minimum = min(age_list)
print("Minimum Age list:",Minimum)
Maximum = max(age_list)
print("Maximum Age list:", Maximum)
Sum = sum(age_list)
print("Sum of Age list:", Sum)
```

**3. Accessing List Elements:**

```python
# a. Print the first element of name_list.
print("First element of name list:", name_list[0])

# b. Print the last element of name_list.
print("last element of name list:", name_list[-1])

# c. Print the elements from index 2 to index 4 in name_list.
print("Elements from index 2 to index 4 in name list:", name_list[2:5])

# d. Print the elements of name_list in reverse order.
print("Elements of name list in reverse order:", name_list[::-1])
```

**Dictionary (Creation, Modification and Access):**

```python
# a. Create a dictionary named student_marks that maps the names of five students to their marks (use a scale of 0 to 100).
student_marks = {
    "Kala":78,
    "Mala":89,
    "Sri":98,
    "Vino":56,
    "Raj":65
    }
print("Student-Marks Dictionary:", student_marks)

# b. Access and print the mark of a specific student of your choice.
print("Student Vino Marks:", student_marks["Vino"])

# c. Add a new student, "Janani," with a mark of 80 to the student_marks dictionary.
student_marks["janani"]=80
print("Added a new student:", student_marks)

# d. Update the mark of any one older student to 82.
student_marks["Kala"]=82
print("Updated the mark for Kala:", student_marks)

# e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.
print("Keys of Student Marks Dictionary:", student_marks.keys())
print("Values of Student Marks Dictionary:", student_marks.values())
print("Key-value pairs of Student Marks Dictionary:", student_marks.items())
```

**Sets (Operations):**

```python
# Sets (Operations):
# a. Create a set called my_set with the following values: ['a','e','i','o','u','a','a','i']
my_set = {'a','e','i','o','u','a','a','i'}
print("My sets:", my_set)
```

**Analyze the output and explain the same**

Output

The output will contain only unique values:
{'o', 'e', 'a', 'i', 'u')

Why are the duplicate values removed?
A set does not allow duplicate elements.

My original values are:
a, e, i, o, u, a, a, i

Here:

'a' appears 3 times → kept only once,
'i' appears 2 times → kept only once,
'e', 'o', 'u' appear once

So the set contains:

{'a', 'e', 'i', 'o', 'u'}

✅ Important: A set stores unique elements only and does not maintain a guaranteed order. Therefore, the order displayed by print() may vary.

**If the code throws an error, provide an explanation**

```python
# b. Attempt to access my_set[4]. If the code throws an error, provide an explanation.
print(my_set[4])
```

**Explanation**

Output/Error:
TypeError: 'set' object is not subscriptable

Explanation:
A set does not support indexing.
This is because sets are unordered collections, so you cannot access their elements using positions like [0], [1], [4], etc.

✅ Conclusion: my_set[4] is invalid because sets do not support indexing.

```python
# c. Create two sets:
# set1 with values: {1, 3, 5, 7, 9}
# set2 with values: {2, 3, 5, 8, 10}
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("Set1:", set1)
print("Set2:", set2)

# d. Compute and print the union and intersection of set1 and set2.
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

**Operators & Conditional Statements : (IF, ELIF, ELSE)**

```python
# Performance Category Program:
# 1. Prompt the user for input. Score range should be from 0 to 10 (both inclusive).
# 2. Find the performance category based on the input score using the following criteria:
# a. Above Average: Score greater than 7
# b. Average: Score between 4 and 7 (both inclusive)
# c. Below Average: Score less than 4
# 3. Output: Print the Performance category
# 4. Additional Step: You can give a prompt of your choice to each category. For example, if the score is below average, “Need to improve your performance; consistent practice will lead to better results”.


Score = int(input("Enter your Score (0 to 10):"))
if Score>=0 and Score<=10:
  if Score>7:
    Staus = "Above Average"
    print(Staus+": Excellent performance! Keep up the good work.")
  elif Score>=4 and Score<=7:
    Staus = "Average"
    print(Staus+": Good effort! Keep practicing to improve your performance.")
  else:
    Staus = "Below Average"
    print(Staus+": Need to improve your performance; consistent practice will lead to better results.")
else:
  print("Your Score range should be between 0 to 10")
```
